## Task Overview
The objective of this task was to identify duplicate records in a dataset, document the findings, and create a cleaned copy of the data. I used the Dirty Customers dataset containing 11 columns, including customer_id, full_name, email, age, order_amount, and account_status.
The final deliverables for this task are a duplicate report and a cleaned copy of the dataset. This exercise strengthened my practical skills in data quality assessment, duplicate detection, and data cleaning using Pandas and Excel.

## Loading Dataset

```python
import pandas as pd
```

```python
df = pd.read_csv("dirty_customers.csv")
print("Originall Shape of Dataset",df.shape)
df.head()
```

## Count Full Row Duplicate Records

```python
duplicate_count = df.duplicated().sum()
print("Total Rows of dataset :",len(df))
print("Full-row Duplictates ",duplicate_count)
print("Unique Rows :",len(df)-duplicate_count)
```

## Display Duplicate Records

```python
duplicates = df[df.duplicated(keep=False)]
print("Rows involved in duplicate groups :",len(duplicates))
display(duplicates.head(20))
```

## Cleaned Copy

```python
cleaned_df = df.drop_duplicates().copy()

print("Rows before cleaning:", len(df))
print("Rows after cleaning:", len(cleaned_df))
print("Rows removed:", len(df) - len(cleaned_df))
print("Remaining full-row duplicates:",
      cleaned_df.duplicated().sum())
```

```python
# Export the duplicate report
duplicates.to_csv("duplicate_report.csv", index=False)

# Export the cleaned dataset
cleaned_df.to_csv("cleaned_superstore.csv", index=False)
```

1. What is a duplicate row?

A duplicate row is a record in a dataset that has the same values as another row in all columns or in the selected key columns.
2. How can duplicates affect totals?
